```python
import numpy as np
import pandas as pd
```

```python
cars_data = pd.read_csv("/content/Toyota.csv",index_col=0,na_values = ["??","????"])
```

```python
cars_data.insert(10,"cars_price","")
```

```python
cars_data.head()
```

```python
for i in range(0,len(cars_data["Price"]),1):
  if (cars_data["Price"][i]<=8450):
    cars_data["cars_price"][i]="low";
  elif(cars_data["Price"][i]>11950):
    cars_data["cars_price"][i]="high";
  else:
    cars_data["cars_price"][i]="medium";
```

```python
i=0
while i<len(cars_data["Price"]):
  if (cars_data["Price"][i]<=8450):
    cars_data["cars_price"][i]="low";
  elif(cars_data["Price"][i]>11950):
    cars_data["cars_price"][i]="high";
  else:
    cars_data["cars_price"][i]="medium";
  i=i+1
```

```python
cars_data.insert(11,"Age_converted",0)
```

```python
def c_converted(val):
  converted = val/12
  return converted
```

```python
cars_data['Age_converted'] = c_converted(cars_data['Age'])
cars_data['Age_converted'] = round(cars_data['Age'],1)
```

```python
cars_data.insert(12,'km_per_month',0)
```

```python
def c_converted(val1,val2):
  val_converted = val1/12
  ratio         = val2/val1
  return[val_converted,ratio]
```

```python
cars_data['Age_converted'],cars_data['km_per_month'] = c_converted(cars_data['Age'],cars_data['KM'])
```
