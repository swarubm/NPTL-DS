```python
import pandas as pd
```

```python
cars = pd.read_csv('/content/Toyota.csv',index_col=0,na_values=["??","????"])
cars_data = cars.copy()
```

identifying null values

```python
cars_data.isnull().sum()
```

we have to check in the same row the values are missing or in the different rows other

```python
missing = cars_data[cars_data.isnull().any(axis=1)]
```

```python
missing
```

fill missing values with mean in numerical and categorical with most count

```python
cars_data.describe()
```

```python
cars_data['Age'].mean()
```

```python
cars_data['Age'].fillna(cars_data['Age'].mean(),inplace=True)
```

```python
cars_data
```

```python
cars_data['KM'].median()
```

```python
cars_data['KM'].fillna(cars_data['KM'].median(),inplace=True)
```

```python
cars_data["HP"].mean()
```

```python
cars_data["HP"].fillna(cars_data["HP"].mean(),inplace=True)
```

```python
cars_data.isnull().sum()
```

```python
cars_data['FuelType'].value_counts()
```

as this is a categorical value we are replaceing with most occured term so in our particular dataset is the more oftenly repeated term thats why we are going to replace missing values with petrol

```python
cars_data['FuelType'].value_counts().index[0]
```

```python
cars_data['FuelType'].fillna(cars_data['FuelType'].value_counts().index[0],inplace=True)
```

```python
cars_data['MetColor'].mode()
```

```python
cars_data['MetColor'].fillna(cars_data['MetColor'].mode()[0],inplace=True)
```
