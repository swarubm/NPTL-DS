```python
cars_data = pd.read_csv("/content/Toyota.csv",index_col=0,na_valu
```

```python
cars_data2=cars_data.copy()     #we have copy of data
```

# Frequency table

```python
pd.crosstab(index=cars_data2["FuelType"],columns="count",dropna=True)
#
```

# Two way table

```python
pd.crosstab(index=cars_data2["Automatic"],columns=cars_data2["FuelType"],dropna=True)
```

# 2 way table joint probability

```python
pd.crosstab(index=cars_data2["Automatic"],columns=cars_data2["FuelType"],dropna=True,normalize=True)
```

```python
pd.crosstab(index=cars_data2["Automatic"],columns=cars_data2["FuelType"],dropna=True,normalize='index')
```

```python
pd.crosstab(index=cars_data2["Automatic"],columns=cars_data2["FuelType"],dropna=True,normalize='columns')
```

# corr

```python
numerical_data = cars_data2.select_dtypes(exclude=[object])
```

```python
print(numerical_data.shape)
```

```python
corr_matrix = numerical_data.corr()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))  # Optional: adjust the figure size
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")  # annot=True adds correlation values
plt.title("Correlation Matrix Heatmap")
plt.show()
```
