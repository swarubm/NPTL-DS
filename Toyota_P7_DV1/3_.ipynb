# 1 Matplotlib

#  a)scatterplot

# it can be used convey the relationship between two numerical variables

#it sometimes helps us to find the correlation b/w the 2 numerical variables

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
cars_data = pd.read_csv("/content/Toyota.csv",index_col=0,na_values=["??","????"])
```

we should drop the missing values so we can easily plot graphs

```python
cars_data.dropna(axis=0,inplace=True)
```

```python
plt.scatter(cars_data["Age"],cars_data["Price"])   #(cars_data["Age"],cars_data["Price"])  as (x,y)
plt.xlabel("Age(months)")
plt.ylabel("Price(euros)")
```

# 2) Histogram

#It is used when we are about to represent the frequency distribution of the variable

```python
plt.hist(cars_data["KM"])
```

#  * To see the graph in a better way we should give color,edge color of the bar and also we can declare bins to how much we want

```python
plt.hist(cars_data["KM"],color="red",edgecolor="black",bins=8)
```

# I will add features or ranges what they are really

```python
plt.hist(cars_data["KM"],color="red",edgecolor="black",bins=8)
plt.xlabel("KM")
plt.ylabel("Frequency")
plt.title("Histogram of KM")
plt.show()
```

# C) Barplot
# it is used to represent categorical data/variables



#It is used to represent freaquency distribution of a categorical variable
#A bar diagram makes easy to compare sets of data between different groups

```python
cars_data['FuelType'].value_counts()
```

```python
count = [968,116,12]
FuelType = ('Petrol','Diesel','CNG')
index = np.arange(len(FuelType))
```

```python
plt.bar(index,count)
plt.xlabel("FuelType")
plt.ylabel("Frequency")
plt.xticks(index,FuelType,rotation=45)
plt.show()
```
